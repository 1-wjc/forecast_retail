```python
import matplotlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option('display.max_columns', None)
# 한글 폰트 설정 (Windows의 경우, 나눔 고딕 폰트를 사용)
plt.rcParams['font.family'] = 'NanumGothic'

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
rjfo = pd.read_csv('../data/(1) 월별 순판매량.csv',encoding="cp949")

# # 중분류 라면 추출 및 필요 없는 열 제거 
# 1데이터
rjfo = rjfo[rjfo['중분류'] == '라면,통조림,상온즉석']
# 2데이터
#rjfo = rjfo[rjfo['대분류'] == '면류.라면류']


rjfo = rjfo.drop(['판매수량','반품수량','월','년'],axis='columns')
rjfo.head()
```

```python
# 원래는 외부 변수만 모은 "유튜브도합친거.csv"를 읽었음. 같은 변수가 든 분석용 변수표에서 판매량 열만 빼고 사용
a = pd.read_csv("../data/최종 변수.csv").drop(columns=["1_순판매량", "2_순판매량"])

# 날짜 열에서 앞의 7자만 남기기
a['날짜'] = a['날짜'].str[:7]

# '날짜' 열에서 앞의 4글자가 '2024'인 행 삭제
a = a[~a['날짜'].str[:4].eq('2024')]


a.head()
```

```python
# '날짜'를 인덱스로 설정 (필요시 인덱스 설정)
rjfo.set_index('날짜', inplace=True)
a.set_index('날짜', inplace=True)

# 두 데이터프레임을 concat으로 합치기
data = pd.concat([rjfo, a], axis=1)

# 인덱스를 일반 열로 변환
data.reset_index(inplace=True)

# 1데이터
data = data.drop(['중분류','날짜'],axis='columns')

# 2데이터
#data = data.drop(['대분류','날짜'],axis='columns')


data.head()
```

* 데이터 에서 NA값을 0으로 변경

```python
# 전체 데이터에서 NA 값을 0으로 변경
data = data.fillna(0)

```
